## 4. Descriptive Statistics: Numerical & Table Summary

Statistik deskriptif membantu merangkum dan memahami karakteristik umum dalam data, baik berupa angka (numerik) maupun kategori.

Kita akan membahas:
- Ukuran pemusatan: mean, median, modus
- Ukuran penyebaran: standar deviasi, IQR
- Frekuensi dan distribusi data kategorikal
- Tabel silang (crosstab) untuk eksplorasi awal dua variabel kategorikal


### 4.1 Ringkasan Statistik Numerik

```python
# TODO: Hitung ukuran pemusatan dan penyebaran data

df[["age", "hrs1"]].describe()
```

- Rata-rata usia responden adalah sekitar `...` tahun.
- Usia tersebar dari `min` hingga `max`, dengan standar deviasi `...`.
- Jam kerja mingguan (`hrs1`) memiliki persebaran besar – kemungkinan karena pensiunan atau part time.

### 4.2 Ukuran Persebaran (posisi): Quartile, IQR

```python
# Hitung IQR
iqr_age = df["age"].quantile(0.75) - df["age"].quantile(0.25)
iqr_hrs1 = df["hrs1"].quantile(0.75) - df["hrs1"].quantile(0.25)

print("IQR usia: ", iqr_age)
print("IQR jumlah jam kerja dalam seminggu: ", iqr_hrs1)
```

```python
# Outlier age

Q3 = df["hrs1"]. quantile(0.75)
Q1 = df["hrs1"]. quantile(0.25)
IQR= Q3 - Q1

above = Q3 + (1.5 * IQR)
bottom = Q1 - (1.5 * IQR)
outlier_hrs1=df[(df["hrs1"] > above) | (df["hrs1"] < bottom)].dropna(subset="hrs1")
outlier_pct1 = len(outlier_hrs1)/len(df)

print(f"Total data outlier: {len(outlier_hrs1)} - {outlier_pct1 :.2%}")
print(f"Batas atas : {above.round()}")
print(f"Batas bawah : {bottom.round()}")
outlier_hrs1.describe()
```

2. Apakah usia seseorang memiliki keterkaitan dengan konteks pekerjaan

```python
# Membandingkan rata-rata umur dari tiap pekerjaan

# Siapkan datanya ! Hitung rata-rata umur dari tiap pekerjaan
age_per_wrkstat = df.groupby(by="wrkstat")["age"].mean().round().sort_values(ascending=True)

# Visualisasikan menggunakan bar chart
plt.barh(y=age_per_wrkstat.index, width=age_per_wrkstat)
plt.title("Rata-rata Usia Responden berdasarkan pekerjaan")
```
